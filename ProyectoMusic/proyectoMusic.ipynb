```python
import pandas as pd
```

```python
# 1. Cargar el CSV con las columnas exactas de Apple
df = pd.read_csv("./Apple Music Play Activity.csv", low_memory=False)

# Columnas reales verificadas del dataset
cols_exactas = [
    "Event Start Timestamp",
    "Song Name",
    "Container Artist Name",  # <--- Este es el nombre real del artista en Apple
    "Album Name",
    "Play Duration Milliseconds",
    "End Reason Type"
]
```

```python
# Filtrar y renombrar para trabajar más cómodo
df_musica = df[cols_exactas].dropna(subset=["Song Name"]).copy()
df_musica = df_musica.rename(columns={"Container Artist Name": "Artist Name"})
```

```python
# Convertir tiempo a minutos y fechas a datetime
df_musica["duracion_min"] = (df_musica["Play Duration Milliseconds"] / (1000 * 60)).round(2)
df_validas = df_musica[df_musica["duracion_min"] >= 0.17].copy()  # Reproducciones > 10s
# df_validas["fecha_hora"] = pd.to_datetime(df_validas["Event Start Timestamp"])
```
