## 2. Listas

Una lista es una secuencia ordenada y modificable. Los índices empiezan en **0**; los negativos cuentan desde el final. El *slicing* `lista[a:b]` toma desde `a` hasta `b` **sin incluir** `b`.

```python
notas = [80, 65, 92, 70, 88]
print(notas[0], notas[-1])       # primero y último
print(notas[1:3])                # posiciones 1 y 2
print(notas[:2], notas[3:])      # los dos primeros; del tercero en adelante
notas.append(55)                 # agrega al final
print(len(notas), sum(notas), min(notas), max(notas))
print(sorted(notas))             # nueva lista ordenada (la original no cambia)
print(92 in notas, 100 in notas) # pertenencia
```

## 3. Diccionarios

Un diccionario guarda pares **clave → valor**. Se accede por clave, no por posición. `d.get(clave, valor_por_defecto)` evita el error cuando la clave no existe; `d.items()` recorre pares. El patrón "contar con un diccionario" es el más importante de esta sección: lo vas a usar en el reto final y es la idea detrás de `groupby` en pandas.

```python
alumno = {"nombre": "Ana", "edad": 18, "carrera": "Ciencia de Datos"}
print(alumno["nombre"], alumno.get("promedio", "sin dato"))
alumno["promedio"] = 82.3            # agregar o modificar
for clave, valor in alumno.items():
    print(clave, "->", valor)

# contar con un diccionario
palabras = ["dato", "modelo", "dato", "dato", "modelo", "grafo"]
conteo = {}
for p in palabras:
    conteo[p] = conteo.get(p, 0) + 1
print(conteo)
```

## 4. Bucles, condicionales y comprensiones

`for` recorre cualquier secuencia; `range(n)` genera 0…n-1; `enumerate` da posición y valor; `zip` recorre dos listas a la vez. Una **comprensión de listas** escribe en una línea el patrón "crear lista vacía, recorrer, agregar": `[expresion for x in secuencia if condicion]`.

```python
for i, n in enumerate([80, 65, 92]):
    print(i, n)

nombres_lista = ["Ana", "Luis", "María"]
notas_lista = [85, 70, 95]
for persona, nota in zip(nombres_lista, notas_lista):
    if nota >= 90:
        print(persona, "excelente")
    elif nota >= 60:
        print(persona, "aprueba")
    else:
        print(persona, "reprueba")

cuadrados = [n * n for n in range(5)]
pares = [n for n in range(10) if n % 2 == 0]
print(cuadrados, pares)
mayusculas = {n: n.upper() for n in nombres_lista}   # comprensión de diccionario
print(mayusculas)
```

## 6. Funciones

Una función encapsula un cálculo para reutilizarlo. Tiene parámetros (con valores por defecto opcionales), un *docstring* que la documenta y `return`. Una `lambda` es una función pequeña de una línea, útil como `key` en `sorted` o `max`.

```python
def promedio_lista(valores, decimales=2):
    """Devuelve el promedio de una lista de números, redondeado."""
    return round(sum(valores) / len(valores), decimales)

print(promedio_lista([85, 70, 92]), promedio_lista([85, 70, 92], 0))

cuadrado = lambda x: x * x
print(cuadrado(7))

pares_nombre_nota = [("Ana", 85), ("Luis", 70), ("María", 95)]
print(sorted(pares_nombre_nota, key=lambda par: par[1]))          # ordenar por nota
print(max(pares_nombre_nota, key=lambda par: par[1]))             # el par con mayor nota
```

## 7. Archivos: CSV y JSON con Python puro

Un **CSV** es texto con una fila por línea y columnas separadas por comas; el módulo `csv` lo lee fila por fila y `csv.DictReader` devuelve cada fila como un diccionario (clave = nombre de columna). Todo lo que se lee de un CSV llega como **texto**: los números hay que convertirlos con `int()` o `float()`.
