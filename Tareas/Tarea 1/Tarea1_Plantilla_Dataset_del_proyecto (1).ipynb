### 5.2 Duplicados y columnas vacías (5 puntos)

*Elimine las filas duplicadas exactas y las columnas totalmente vacías. Si su tabla no tiene identificador de fila, decida con el diccionario de datos si dos filas iguales pueden ser reales.*

```python
limpio = datos.drop_duplicates().reset_index(drop=True)
vacias = limpio.columns[limpio.isna().all()].tolist()
limpio = limpio.drop(columns=vacias)

print("Filas eliminadas:", len(datos) - len(limpio), "| columnas eliminadas:", vacias)
limpio.shape
```

*Decisión:*Se eliminaron las filas duplicadas exactas y las columnas totalmente vacías. En este dataset no se encontraron filas duplicadas ni columnas completamente vacías, por lo que no se eliminó ningún registro ni ninguna columna. Se conservaron las 1000 filas y las 16 columnas originales.

### 5.3 Texto: espacios, mayúsculas y sinónimos (5 puntos)

*Una celda por columna de texto con variantes. El orden importa: espacios → caja → sinónimos (`replace`). Al final, cada columna debe tener una sola forma de escribir cada valor.*

```python
limpio["Certificate"] = (limpio["Certificate"].str.strip().replace({"": pd.NA}))
limpio["Certificate"].value_counts(dropna=False)
```

```python
columnas_texto = [
    "Genre", "Director",
    "Star1", "Star2", "Star3", "Star4"
]

for col in columnas_texto:
    limpio[col] = limpio[col].str.strip()

limpio[columnas_texto].nunique()
```

*Decisión (qué variantes se unificaron y por qué): Se eliminaron los espacios sobrantes al inicio y al final de los valores de las columnas de texto, como Certificate, Genre, Director, Star1, Star2, Star3 y Star4. Se conservaron las mayúsculas, minúsculas y la escritura original de los títulos y nombres propios para evitar modificar información válida. No se reemplazaron sinónimos porque no se identificaron variantes que necesitaran unificarse.

### 5.4 Tipos: números y fechas (5 puntos)

*Convierta a número las columnas numéricas leídas como texto (`str.replace` para quitar símbolos, `pd.to_numeric(errors="coerce")`) y a fecha las columnas de fecha (`pd.to_datetime` con `format=`; un `to_datetime` por formato si hay varios). Compruebe después que `dtypes` muestra lo que dice el diccionario de datos.*

```python

# 1. Año de estreno
limpio["Released_Year"] = pd.to_numeric(limpio["Released_Year"].astype("string").str.strip(),errors="coerce").astype("Int64")

# 2. Duración en minutos
limpio["Runtime"] = pd.to_numeric(limpio["Runtime"].astype("string").str.replace(" min", "", regex=False).str.strip(),errors="coerce").astype("Int64")

# 3. Recaudación en dólares
limpio["Gross"] = pd.to_numeric(limpio["Gross"].astype("string").str.replace("$", "", regex=False).str.replace(",", "", regex=False).str.strip(),errors="coerce").astype("Int64")

print(limpio[["Released_Year", "Runtime", "Gross"]].dtypes)
print("\nValores faltantes:")
print(limpio[["Released_Year", "Runtime", "Gross"]].isna().sum())

```

*Columnas convertidas y motivo:* Released_Year: se convirtió de string a entero (Int64) porque representa el año de estreno de cada película. Los valores inválidos se transformaron en faltantes para evitar errores.

Runtime: se convirtió de string a entero (Int64) porque representa la duración en minutos. Se eliminó el texto " min" antes de la conversión.

Gross: se convirtió de string a entero (Int64) porque representa la recaudación en dólares estadounidenses. Se eliminaron las comas separadoras de miles y, si estaba presente, el símbolo $.

Se utilizó errors="coerce" para convertir los valores que no se pueden interpretar como números en valores faltantes (NA). Se conservaron las filas originales.

### 5.6 Valores atípicos (4 puntos)

*Revise `describe()` y la regla IQR en las columnas numéricas. Corrija solo los errores evidentes (y documente la corrección); un valor extremo real se queda.*

```python
limpio.describe()
```

```python

columna = "Runtime"
q1, q3 = limpio[columna].quantile([0.25, 0.75])
iqr = q3 - q1
limite_inf = q1 - 1.5 * iqr
limite_sup = q3 + 1.5 * iqr
atipicos = limpio[(limpio[columna] < limite_inf) |(limpio[columna] > limite_sup)]

print("Q1:", q1)
print("Q3:", q3)
print("IQR:", iqr)
print("Límite inferior:", limite_inf)
print("Límite superior:", limite_sup)
print("Cantidad de valores atípicos:", len(atipicos))

atipicos[["Series_Title", columna]].sort_values(
    columna, ascending=False
)
```

```python
limpio[["Series_Title", "Runtime"]].sort_values(
    "Runtime", ascending=False
).head(10)
```

*Decisión (qué se corrigió, qué se dejó y por qué):* Se revisaron los valores extremos de las columnas numéricas mediante estadísticas descriptivas y la regla del rango intercuartílico (IQR). Se analizaron las películas con mayor duración, como *Gangs of Wasseypur* (321 minutos) y *Hamlet* (242 minutos). No se modificaron ni eliminaron estos valores, ya que no se confirmó que fueran errores. También se decidió conservar los valores extremos de `No_of_Votes`, `IMDB_Rating` y `Meta_score`, porque pueden representar datos reales importantes para el análisis de popularidad y calificaciones. *Pregunta 1:* ¿Cuáles son las 5 películas mejor valoradas (con mayor IMDB_Rating) y cuáles son sus directores y años de estreno?

```python
# selección 1
top_5_rating = limpio.sort_values(by="IMDB_Rating", ascending=False).loc[:, ["Series_Title", "Released_Year", "Director", "IMDB_Rating"]].iloc[:5]
top_5_rating
```

*Interpretación:* Las 5 películas con la calificación más alta en la plataforma destacan por superar los 8.9 puntos de calificación IMDb, reflejando las producciones mejor evaluadas por el público en el dataset.

*Pregunta 2:* ¿Qué películas pertenecen al género de "Drama" y tienen más de 500,000 votos (No_of_Votes)?

```python
# selección 2
drama_populares = limpio.loc[(limpio["Genre"].str.contains("Drama", na=False)) & (limpio["No_of_Votes"] > 500000), ["Series_Title", "Released_Year", "IMDB_Rating", "No_of_Votes"]]
drama_populares.head()
```

*Interpretación:* La selección filtra las películas de Drama que han alcanzado una alta popularidad, permitiendo identificar los títulos del género dramático con mayor alcance e interacción de la audiencia.

*Pregunta 3:* ¿Cuáles son las primeras 10 películas en el conjunto de datos que fueron estrenadas a partir del año 2010?

```python
# selección 3
recientes_pos2010 = limpio.loc[limpio["Released_Year"].astype(str) >= "2010", ["Series_Title", "Released_Year", "Genre", "IMDB_Rating"]].iloc[:10]
recientes_pos2010
```

*Interpretación:* Esta consulta permite aislar un bloque de producciones recientes (estrenadas de 2010 en adelante) para analizar la presencia del cine moderno dentro del Top 1000 de IMDb.
