## 1. De NumPy a pandas: por qué hacen falta etiquetas

Estos son los datos de la sesión 4: las notas de seis estudiantes en tres evaluaciones.

```python
nombres = np.array(["Ana", "Luis", "María", "Jorge", "Camila", "Diego"])
notas = np.array([[85, 90, 78],
                  [70, 65, 80],
                  [95, 92, 98],
                  [55, 60, 58],
                  [88, 79, 91],
                  [62, 74, 69]])

notas[4, 2]     # ¿de quién es esta nota y de qué evaluación? El array no lo dice
```

Con un array hay que **recordar** que la fila 4 es Camila y que la columna 2 es la tercera evaluación. Además, los nombres viven en otro array, porque un array solo admite un tipo de dato.

Un **DataFrame** de pandas guarda la misma matriz junto con sus etiquetas: un **índice** (*index*) para las filas y un nombre para cada **columna**.

```python
df_notas = pd.DataFrame(notas, index=nombres, columns=["ev1", "ev2", "ev3"])
df_notas
```

```python
df_notas.loc["Camila", "ev3"]     # la misma nota, ahora pedida por sus etiquetas
```

```python
print(df_notas.shape)      # (filas, columnas), igual que en NumPy
print(df_notas.index)      # las etiquetas de las filas
print(df_notas.columns)    # las etiquetas de las columnas
```

| | Array de NumPy | DataFrame de pandas |
|---|---|---|
| Se eligen los datos por… | posición: `notas[4, 2]` | etiqueta: `df_notas.loc["Camila", "ev3"]` (y también por posición) |
| Tipos de datos | uno solo para todo el array | uno por columna |
| Sirve para… | cálculo numérico con matrices | tablas de datos: archivos CSV, Excel, bases de datos |

## 2. `Series`: un array con etiquetas

Una **columna** de un DataFrame es una **Series**: un array de una dimensión en el que cada valor tiene una etiqueta.

```python
ev1 = df_notas["ev1"]        # una columna del DataFrame
print(type(ev1))
ev1
```

```python
print(ev1.to_numpy())    # los valores: un array de NumPy
print(ev1.index)         # las etiquetas
print(ev1.name, "|", ev1.dtype, "|", ev1.shape)
```

Una Series también se crea directamente, con una lista y un índice, o con un diccionario (las claves pasan a ser las etiquetas).

```python
edades = pd.Series([19, 21, 20], index=["Ana", "Luis", "María"], name="edad")
ciudad_origen = pd.Series({"Ana": "Guayaquil", "Luis": "Quito", "María": "Cuenca"})
print(edades)
print(ciudad_origen)
```

**Elegir valores.** Por etiqueta se usan corchetes o `.loc`; por posición, `.iloc`.

```python
print(ev1["María"])                 # por etiqueta
print(ev1.iloc[2])                  # por posición (la tercera)
print(ev1[["Ana", "Jorge"]])        # varias etiquetas: una lista dentro de los corchetes
print(ev1["Luis":"Jorge"])          # slice por etiquetas: el fin SÍ se incluye
```

> **Atención.** Con un índice de texto, `ev1[2]` no es una forma segura de pedir «el tercer valor»: en pandas 3 produce un error (`KeyError`), porque el número 2 no es una etiqueta. Para pedir por posición se usa siempre `.iloc`.

**Todo lo de NumPy sigue funcionando**: operaciones vectorizadas, comparaciones, máscaras y agregaciones. La diferencia es que el resultado conserva las etiquetas.

```python
print(ev1 + 5)                      # operación vectorizada: conserva las etiquetas
print(ev1[ev1 >= 80])               # máscara booleana, como en NumPy
print(ev1.mean(), ev1.max(), (ev1 >= 60).sum())
print(ev1.idxmax())                 # como argmax, pero devuelve la ETIQUETA del máximo
```

**Las operaciones entre dos Series se alinean por etiqueta, no por posición.** Si una etiqueta falta en una de las dos, el resultado es `NaN` (*not a number*): el valor que pandas usa para un dato faltante. Los datos faltantes se estudian en la sesión 7.

```python
bono = pd.Series({"Diego": 5, "Ana": 10})    # solo dos estudiantes, y en otro orden
ev1 + bono
```

## 5. Filas: `loc` e `iloc`

Hay dos formas de elegir filas (y columnas), y conviene no mezclarlas:

| | Elige por… | En un *slice*, el fin… | Ejemplo |
|---|---|---|---|
| `df.iloc[filas, columnas]` | **posición** (enteros desde 0), como NumPy | **no** se incluye | `ventas.iloc[0:3, [1, 4]]` |
| `df.loc[filas, columnas]` | **etiqueta** (del índice y de las columnas) | **sí** se incluye | `ventas.loc[0:3, ["producto", "ingreso"]]` |

Con `df_notas` la diferencia se ve bien, porque las etiquetas de las filas son nombres:

```python
print(df_notas.loc["María"])                        # la fila con etiqueta "María" → Series
print(df_notas.iloc[2])                             # la fila en la posición 2: la misma
```

```python
print(df_notas.loc["Luis":"Jorge", "ev1":"ev2"])    # por etiquetas: incluye a Jorge y a ev2
print(df_notas.iloc[1:4, 0:2])                      # por posiciones: el 4 y el 2 quedan fuera
```

En `ventas`, las etiquetas de las filas son los números 0, 1, 2, …, y por eso `loc` e `iloc` **parecen** iguales. No lo son:

```python
print(len(ventas.iloc[0:3]), "filas con iloc[0:3]")     # posiciones 0, 1 y 2
print(len(ventas.loc[0:3]), "filas con loc[0:3]")       # etiquetas 0, 1, 2 y 3
ventas.loc[0:3, ["producto", "ingreso"]]
```

La diferencia importa en cuanto la tabla se **ordena** o se **filtra**: las posiciones cambian, las etiquetas no.

```python
orden = ventas.sort_values("ingreso", ascending=False)

print(orden.iloc[0]["producto"], orden.iloc[0]["ingreso"])   # posición 0: la venta de MAYOR ingreso
print(orden.loc[0]["producto"], orden.loc[0]["ingreso"])     # etiqueta 0: la PRIMERA venta del archivo
```

Un valor suelto se pide con fila y columna: `ventas.loc[57, "producto"]`.

El índice se puede cambiar: `set_index` convierte una columna en índice, y `reset_index` hace lo contrario.

```python
tabla = df_notas.reset_index()             # el índice pasa a ser una columna llamada "index"
print(tabla.head(2))
print(tabla.set_index("index").head(2))    # y vuelve a ser el índice
```

## 6. Condiciones

Una comparación sobre una columna devuelve una **Series de `True`/`False`**: una máscara, igual que en NumPy. Con ella se cuenta, se calcula una proporción y se filtran filas.

```python
en_guayaquil = ventas["ciudad"] == "Guayaquil"
print(en_guayaquil.head(4))
print(en_guayaquil.sum(), "ventas en Guayaquil")            # contar: 76
print(en_guayaquil.mean() * 100, "% de las ventas")         # proporción: 38 %
```

```python
ventas[en_guayaquil].head(3)                                # filtrar filas: solo las de Guayaquil
```

Con `.loc` se eligen **filas por condición y columnas por nombre** en un solo paso: `df.loc[condición, columnas]`.

```python
ventas.loc[en_guayaquil, ["producto", "ingreso"]].head(3)
```

```python
ventas.loc[ventas["categoria"] == "Electrónica", "ingreso"].sum()      # 3397.5, como en el Taller 1
```

Para **combinar condiciones** se usan `&` (y), `|` (o) y `~` (no), **con paréntesis en cada condición**. Igual que con los arrays, `and` y `or` no funcionan.

```python
quito_grandes = (ventas["ciudad"] == "Quito") & (ventas["cantidad"] >= 5)
print(quito_grandes.sum(), ventas.loc[quito_grandes, "ingreso"].sum())     # 16 ventas, 1537.3

# ventas[ventas["ciudad"] == "Quito" and ventas["cantidad"] >= 5]
# ValueError: The truth value of a Series is ambiguous
```

Dos métodos ahorran condiciones largas: `isin` (el valor está en una lista) y `between` (el valor está en un intervalo, con los extremos incluidos).

```python
print(ventas["ciudad"].isin(["Cuenca", "Manta"]).sum())           # ventas en Cuenca o en Manta
print(ventas["precio_unitario"].between(10, 20).sum())            # precios de 10 a 20 dólares
print((~ventas["ciudad"].isin(["Guayaquil", "Quito"])).sum())     # ~ niega: ni Guayaquil ni Quito
```

**Modificar con una condición.** La forma segura es `df.loc[condición, columna] = valor`.

```python
ventas["tipo"] = "minorista"                                   # un mismo valor para toda la columna
ventas.loc[ventas["cantidad"] >= 5, "tipo"] = "mayorista"      # se cambia solo donde la condición es True
ventas["tipo"].value_counts()
```

> **Atención.** La forma `ventas[ventas["cantidad"] >= 5]["tipo"] = "mayorista"` (dos pares de corchetes seguidos) **no modifica** `ventas`: el primer corchete crea una tabla nueva y el cambio se hace sobre esa tabla temporal. pandas avisa con `SettingWithCopyWarning` (pandas 2) o `ChainedAssignmentError` (pandas 3). Para modificar: siempre un solo `.loc`. Para trabajar con una parte de la tabla como tabla independiente: `parte = ventas[condición].copy()`.

## 7. Tipos de datos y regreso a NumPy

El tipo de una columna decide qué operaciones se pueden hacer con ella. `fecha` se leyó como texto: así no se puede pedir el mes ni restar dos fechas. `pd.to_datetime` la convierte en una fecha de verdad.

```python
print(ventas["fecha"].dtype)                       # texto
ventas["fecha"] = pd.to_datetime(ventas["fecha"])
print(ventas["fecha"].dtype)                       # datetime64
```

```python
print(ventas["fecha"].min(), "→", ventas["fecha"].max())
print(ventas["fecha"].dt.month.head(3).tolist())            # .dt da acceso a las partes de la fecha
print(ventas["fecha"].dt.dayofweek.head(3).tolist())        # día de la semana: 0 = lunes … 6 = domingo
print((ventas["fecha"] >= "2026-03-01").sum(), "ventas en marzo")
```

> También se puede convertir al leer el archivo: `pd.read_csv("datos/ventas_tienda.csv", parse_dates=["fecha"])`.

`astype` cambia el tipo de una columna, como en NumPy. El tipo `category` es útil para variables categóricas con pocos valores distintos.

```python
print(ventas["cantidad"].astype(float).head(3).tolist())
categoria = ventas["categoria"].astype("category")
print(categoria.cat.categories.tolist())
```

**Un dato escrito con dígitos no siempre es un número.** En la sesión 3 se clasificó el número de cédula como categórico nominal: es un identificador. Si pandas lo lee como número, se pierde el cero inicial (las cédulas de las provincias 01 a 09 empiezan con 0; la de Guayas, con 09). Los números de este ejemplo son ficticios.

```python
import io

texto = "cedula,nombre\n0911111111,Ana\n1722222222,Luis\n0133333333,María\n"

mal = pd.read_csv(io.StringIO(texto))                              # pandas decide: número entero
bien = pd.read_csv(io.StringIO(texto), dtype={"cedula": str})      # se indica: texto
print(mal["cedula"].tolist())
print(bien["cedula"].tolist())
```

## Resumen

| Para… | Se escribe | Ejemplo |
|---|---|---|
| leer un archivo | `pd.read_csv` | `ventas = pd.read_csv("datos/ventas_tienda.csv")` |
| inspeccionar | `.head()`, `.tail()`, `.shape`, `.columns`, `.dtypes`, `.info()`, `.describe()` | `ventas.dtypes` |
| elegir una columna | `df["col"]` → Series | `ventas["ciudad"]` |
| elegir varias columnas | `df[["a", "b"]]` → DataFrame | `ventas[["producto", "ciudad"]]` |
| crear una columna | `df["nueva"] = …` | `ventas["ingreso"] = ventas["cantidad"] * ventas["precio_unitario"]` |
| elegir por posición | `df.iloc[filas, columnas]` (fin excluido) | `ventas.iloc[:5, :3]` |
| elegir por etiqueta | `df.loc[filas, columnas]` (fin incluido) | `ventas.loc[10:12, ["producto", "ciudad"]]` |
| filtrar por condición | `df[condición]`, `df.loc[condición, columnas]` | `ventas.loc[ventas["ciudad"] == "Quito", "ingreso"]` |
| combinar condiciones | `&`, `\|`, `~`, `.isin()`, `.between()` | `(ventas["ciudad"] == "Quito") & (ventas["cantidad"] >= 5)` |
| modificar con condición | `df.loc[condición, "col"] = valor` | `ventas.loc[ventas["cantidad"] >= 5, "tipo"] = "mayorista"` |
| resumir una columna | `.sum()`, `.mean()`, `.max()`, `.idxmax()`, `.value_counts()`, `.nunique()` | `ventas["ciudad"].value_counts()` |
| ordenar | `.sort_values()` | `ventas.sort_values("ingreso", ascending=False)` |
| cambiar tipos | `pd.to_datetime`, `.astype()` | `ventas["fecha"] = pd.to_datetime(ventas["fecha"])` |
| pasar a NumPy | `.to_numpy()` | `ventas[["cantidad", "ingreso"]].to_numpy()` |

### Errores frecuentes

| Síntoma | Causa | Solución |
|---|---|---|
| `KeyError: 'Ciudad'` | el nombre de la columna no existe (mayúsculas, tildes, espacios) | copiar el nombre de `df.columns` |
| `KeyError: ('producto', 'ciudad')` | varias columnas con corchetes simples | corchetes dobles: `df[["producto", "ciudad"]]` |
| `ValueError: The truth value of a Series is ambiguous` | se usó `and` u `or` | `&` y `\|`, con paréntesis en cada condición |
| Sale una fila de más (o de menos) | se confundió `loc` (fin incluido) con `iloc` (fin excluido) | etiquetas → `loc`; posiciones → `iloc` |
| `SettingWithCopyWarning` o `ChainedAssignmentError` | asignación con dos corchetes seguidos | un solo `df.loc[condición, columna] = valor` |
| Una columna numérica no se puede sumar | se leyó como texto | revisar `df.dtypes`; convertir con `astype` o al leer |
| `FileNotFoundError` | Jupyter se abrió en otra carpeta | abrir Jupyter desde `FCDIA`; revisar `os.getcwd()` |

**Ante cualquier resultado inesperado, revise primero `.shape`, `.dtypes` y `.head()`.**
